## Hands‑on: Interactive Decision Boundaries with Python

Welcome to the programming session companion for Module 2! In this notebook, we will use **Python**, `scikit-learn`, and `matplotlib` to recreate exactly what you observed in the online visual tools.

### Objectives:
1. **Step 1:** Train a simple linear model on an easy, separable dataset.
2. **Step 2:** Watch a linear model struggle (underfit) on a non-linear dataset.
3. **Step 3:** Switch to non-linear models (Decision Trees/KNN) to successfully capture complex shapes.
4. **Step 4:** Push a non-linear model to its limit to observe overfitting (jagged boundaries/noise-chasing).

```python
# ========================================== 
# INITIAL SETUP: Run this cell first!
# Imports the required libraries and helpers
# ==========================================
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_circles
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

def plot_decision_boundary(model, X, y, title):
    """Helper function to plot data points and decision boundaries"""
    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02), np.arange(y_min, y_max, 0.02))
    
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    
    plt.figure(figsize=(8, 6))
    plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.coolwarm, s=40)
    
    # Calculate accuracy
    preds = model.predict(X)
    acc = accuracy_score(y, preds) * 100
    
    plt.title(f"{title} (Training Accuracy: {acc:.1f}%)")
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.show()

print("Setup complete! Ready to start.")
```

--- 
# Step 1: Simple Linear Model
**Goal:** See how a linear model behaves on two clearly separated clusters. This should yield high performance and a perfectly straight boundary line.

```python
# 1. Create a simple 2-class dataset (blobs)
X_simple, y_simple = make_blobs(n_samples=200, centers=2, random_state=42, cluster_std=1.2)

# 2. Select and train a linear model (Logistic Regression)
linear_model = LogisticRegression()
linear_model.fit(X_simple, y_simple)

# 3. Plot the result
plot_decision_boundary(linear_model, X_simple, y_simple, "Step 1: Logistic Regression on Simple Data")
```

# Step 2: Underfitting with Linear Models
**Goal:** See what happens when a linear model is forced onto a complex, non-linear pattern (concentric circles).

```python
# 1. Create a complex 2-class dataset (concentric circles)
X_complex, y_complex = make_circles(n_samples=200, noise=0.1, factor=0.5, random_state=42)

# 2. Keep the same linear model type
underfitting_linear = LogisticRegression()
underfitting_linear.fit(X_complex, y_complex)

# 3. Plot the result
plot_decision_boundary(underfitting_linear, X_complex, y_complex, "Step 2: Linear Model Underfitting Circular Data")
```

### Step 2 Questions for Discussion:
1. Can the linear boundary wrap around the inner circle?
2. Look at the Training Accuracy: Why is it hovering near 50%?
3. Why is this an textbook example of **underfitting**?

# Step 3: Switch to Non‑Linear Models
**Goal:** See how highly flexible, non-linear models (like Decision Trees or K-Nearest Neighbors) can bend boundaries to successfully isolate complex data structures.

```python
# 1. Use the same circular dataset from Step 2
# 2. Initialize a non-linear model (Decision Tree with moderate depth)
nonlinear_tree = DecisionTreeClassifier(max_depth=4, random_state=42)
nonlinear_tree.fit(X_complex, y_complex)

# 3. Plot the result
plot_decision_boundary(nonlinear_tree, X_complex, y_complex, "Step 3: Decision Tree (Max Depth = 4) on Circular Data")
```

# Step 4: Overfitting (Bonus Challenge)
**Goal:** Observe what happens when a model is given *too much* complexity, causing its boundary shapes to become overly jagged while chasing outliers/noise.

```python
# 1. Create a heavily overlapping, noisy dataset
X_noisy, y_noisy = make_circles(n_samples=200, noise=0.25, factor=0.5, random_state=10)

# 2. Train an overly complex Decision Tree (Unrestricted Depth)
overfitting_tree = DecisionTreeClassifier(max_depth=None, random_state=42)
overfitting_tree.fit(X_noisy, y_noisy)

# 3. Plot the result
plot_decision_boundary(overfitting_tree, X_noisy, y_noisy, "Step 4: Overfitting Tree (No Depth Limit)")
```
